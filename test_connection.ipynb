```python
# setup
from pathlib import Path

from config import CONNECTION_PASSWORD, DB_HOST, DB_NAME, DB_PORT, DB_USER
from connection import getDbEngine, loadSqlFromFile, renderSqlTemplate, runSqlToDf
```

```python
# set connection vars
host = DB_HOST
port = DB_PORT
database = DB_NAME
username = DB_USER
password = CONNECTION_PASSWORD

engine = getDbEngine(
    host=host,
    database=database,
    username=username,
    password=password,
    port=port,
)
```

```python
# load and render sql template
sqlPath = Path("./sql_templates/exampleTemplate")

templateVars = {
    "schema_name": "public",
    "table_name": "aca_plan_attributes_puf",
}

sqlTemplate = loadSqlFromFile(sqlPath)
sqlFinal = renderSqlTemplate(sqlTemplate, templateVars=templateVars)

print(sqlFinal)
```

```python
# run query
df = runSqlToDf(engine, sqlFinal)
df
```
